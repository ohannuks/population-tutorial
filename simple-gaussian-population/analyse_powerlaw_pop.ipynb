# Population assumptions

- No redshift evolution

$p(d_L) \propto \text{constant} \times \Theta(d_L<20000)$

- Power-law chirp mass 

$p(\mathcal M) \propto \mathcal M^{-\alpha} $

with $5 \text{M}_\odot <\mathcal M<200 \text{M}_\odot$

- Assume SNR computation as per https://arxiv.org/pdf/gr-qc/0509116 :

Assume:

(1) $D_\text{eff} = D$ is the luminosity distance (fix $\iota=\pi/2$ and $F_+=2$)

(2) Assume mass ratio $m_2/m_1=1$, so that reduced mass $\mu=m^2/2m=m/2$

(3) Assume that the merger happens at high frequency such that

$\rho_\text{opt} = A(D_\text{eff}, \mathcal M) \int \frac{f^{-2*7/6}}{S_n(f)} \mathrm d f \approx A(D_\text{eff}, \mathcal M)\times  c_0$

($c_0$ needs to be computed)

- Selection function from https://arxiv.org/abs/1809.02293 

$$p(\rho_\text{mf} | \rho_\text{opt}) \propto \exp \left[ - \left( \frac{\rho_\text{mf} - \rho_\text{opt}}{2} \right)^2 \right]$$

such that:

$$p_\text{det}(\mathcal M, D_\text{eff}) = \frac{1}{2} \text{erfc} \left (\frac{\rho_\text{thr}-\rho_\text{opt}(\mathcal M, D_\text{eff})}{\sqrt{2}} \right)$$

- Assume one detector

# GW assumptions

GW measurement/likelihood is gaussian:

$$p(d|\log \mathcal M) \propto \exp\left[ -\left(\frac{\log \mathcal M - \log \mathcal M^\text{median}}{2\sigma_{\log \mathcal M}} \right)^2\right]$$

Assume $\sigma_{\log \mathcal M} \approx \log (0.1 \text{M}_\odot)$

(missing the fact that the error on chirp mass depends on the SNR)

# Question:

What is the power-law index $\alpha$?

```python
def A(Mchirp, Deff):
    ''' Calculate the amplitude of the waveform at the detector, i.e., the A in:

    h(f) = A(Mchirp, Deff) * f^{-7/6} * exp(i * Psi(f))

    Parameters
    Mchirp : The chirp mass of the binary in solar masses
    Deff : The effective distance of the binary in Mpc

    Returns
    A : The amplitude of the waveform at the detector
    '''
    return - 1.60894e-21 * Mchirp**(5/6) / Deff
```
